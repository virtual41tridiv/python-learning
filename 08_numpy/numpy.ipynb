## Matrix operations

```python
import numpy as np
```

```python
q1 = np.array([
    [200, 220, 250],
    [150, 180, 210],
    [300, 330, 360]
])

q2 = np.array([
    [209, 231, 259],
    [155, 192, 222],
    [310, 340, 375]
])
```

```python
q1 + q2
```

```python
q2 - q1
```

```python
(q2-q1)*100/q1
```

```python
prices = np.array([
    [10, 12, 11],
    [8, 9, 10],
    [15, 16, 17]
])

q1_revenue = q1 * prices
q1_revenue
```

```python
q1_discounts = q1_revenue*0.2
q1_discounts
```

```python
net_revenue = q1_revenue - (q1_discounts)
```

```python
net_revenue
```

```python
np.sum(net_revenue)
```

```python
# Customer ID
c = np.array([
    [101, 'Mira'],
    [102, 'Ravi'],
    [103, 'Abdul']
])
#Customer ID, Purchase Amount, Purchase Date
d = np.array([
    [101, 250.50, '2023-08-01'],
    [102, 150.00, '2023-08-02'],
    [103, 300.75, '2023-08-01']
])
```

```python
# We will use horizontal stacking to combine the two arrays based on the customer ID. However, since the customer names are strings and the purchase amounts are floats, we need to ensure that the data types are compatible. We can convert the purchase amounts to strings for this purpose.
np.hstack((c,d))
```
