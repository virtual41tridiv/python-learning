## Exception Handling and Operator Overloading

```python
import datetime
```

```python
class cricketplayer:
    teamsize = 11
    def __init__(self, first_name, last_name, scores, birth_year,team):
        self.first_name = first_name
        self.last_name = last_name
        self.scores = scores
        self.birth_year = birth_year
        self.team = team

    def add_score(self, score):
        self.scores.append(score)

    def get_average_score(self):
        return sum(self.scores) / len(self.scores)

    def age(self):
        current_year = datetime.datetime.now().year
        return current_year - self.birth_year

    def __str__(self):
        return f"{self.first_name} {self.last_name}, Team: {self.team}, Average Score: {self.get_average_score()}, Age: {self.age()}"   

    def __lt__(self, other):
        self_average_score = self.get_average_score()
        other_average_score = other.get_average_score()
        return self_average_score < other_average_score
        
    def __eq__(self, other):
        self_age = self.age()
        other_age = other.age()
        return self_age == other_age

virat = cricketplayer('Virat', 'Kohli', [100, 80, 23], 1988, 'India')
print(virat.first_name)
print(virat.last_name)
print(virat.scores)
print(virat.birth_year)
print(virat.get_average_score())
print(virat.age())
print(virat.team)
print(virat)

david = cricketplayer('David', 'Warner', [50, 60, 70], 1986, 'Australia')
print(david.first_name)
print(david.last_name)
print(david.scores)
print(david.birth_year)
print(david.get_average_score())        
print(david.age())  
print(david.team)
print(david)
```

## Inheritance

```python
class Player:
    def __init__(self, first_name, last_name, birth_year):
        self.first_name = first_name
        self.last_name = last_name
        self.birth_year = birth_year

    def get_age(self):
        current_year = datetime.datetime.now().year
        return current_year - self.birth_year


class TennisPlayer(Player):
    def __init__(self, first_name, last_name, birth_year, ranking):
        super().__init__(first_name, last_name, birth_year)
        self.ranking = ranking
        self.aces = []


    def get_average_aces(self):
        return sum(self.aces) / len(self.aces)
    

    def get_ranking(self):
        return self.ranking


class CricketPlayer(Player):
    def __init__(self, first_name, last_name, birth_year, team):
        super().__init__(first_name, last_name, birth_year)
        self.team = team
        self.scores = []

    def get_average_score(self):
        return sum(self.scores) / len(self.scores)

    def get_team(self):
        return self.team    


roger = TennisPlayer('Roger', 'Federer', 1981, 1)
roger.aces.append(10)
roger.aces.append(15)
print(roger.first_name)
print(roger.last_name)
print(roger.birth_year)
print(roger.get_age())
print(roger.get_average_aces())

virat = CricketPlayer('Virat', 'Kohli', 1988, 'India')
virat.scores.append(100)
virat.scores.append(80)
print(virat.first_name)
print(virat.last_name)      



```
